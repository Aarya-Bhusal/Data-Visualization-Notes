# Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Basic Ploting and Labeling**

```python
def sqr_func(arr):
    return arr**2

x = np.linspace(-10,10,50)
y = sqr_func(x)
plt.plot(x, y)
plt.title("Square Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```
