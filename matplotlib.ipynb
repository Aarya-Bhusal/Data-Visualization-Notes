# Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Basic Ploting and Labeling**

```python
def sqr_func(arr):
    return arr**2

x = np.linspace(-10,10,50)
y = sqr_func(x)
plt.plot(x, y)
plt.title("Square Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

**Subplots in matplotlib**

```python
def sqr_func(arr):
    return arr**2

def cube_func(arr):
    return arr**3

def sqr_root_func(arr):
    return np.sqrt(arr)

def cube_root_func(arr):
    return np.cbrt(arr)

x = np.linspace(-100,100,50)

plt.figure(figsize=(25,25))

# First Plot
plt.subplot(2,2,1)
y = sqr_func(x)
plt.plot(x,y)
plt.title("Square Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

# Second Plot
plt.subplot(2,2,2)
y = cube_func(x)
plt.plot(x,y)
plt.title("Cube Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

# Third Plot
plt.subplot(2,2,3)
y = sqr_root_func(x)
plt.plot(x,y)
plt.title("Square Root Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

# Forth Plot
plt.subplot(2,2,4)
y = cube_root_func(x)
plt.plot(x,y)
plt.title("Cube Root Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

**Object Oriented Way of Controling Canvas**

```python
x = np.linspace(1,100, 1000)
y = np.sqrt(x)
fig = plt.figure(figsize=(14,14), dpi=100) #Create a figure on which to plot
ax1 = fig.add_axes([0,0,1,1]) #Create a new figure on the fig canvas. [left-margin, bottom-margin, breadth(0-1), height(0-1)]
ax2 = fig.add_axes([1.2,0,1,1]) #Create a new figure on the fig canvas. [left-margin, bottom-margin, breadth(0-1), height(0-1)]
ax1.plot(x,y)
ax2.plot(y,x)
```

**Saving Plots**

```python
figure = plt.figure(figsize=(10,5), dpi=100)
plot1 = figure.add_axes([0,0,1,1])
plot1.set(
    title="My Plot",
    xlabel="x-axis",
    ylabel="y-axis"
)
x = np.random.randint(1,101,50)
y = np.random.randint(1,101,50)
plot1.scatter(x,y)
figure.savefig("figure.png", bbox_inches="tight") # Keep bbox_inches='tight' to save the axis numbering and labels
```
