# Matrix-Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
df
```

```python
def get_sex_num(sex):
    if sex=="Male":
        return 1
    return 0

def get_smoker_num(smoking_status):
    if smoking_status=="Yes":
        return 1
    return 0

df["sex_num"] = df["sex"].apply(get_sex_num)
df["smoker_num"] = df["smoker"].apply(get_smoker_num)
df
```

**Corelation and Heat Map**

```python
# Corelation tells us how much are numeric columns related to each other
numeric_df = df[["total_bill", "tip", "size", "sex_num", "smoker_num"]] # Only take numeric columns
numeric_df.corr() # Get the corelation
```

```python
# Heatmap is just a visual representation of that corelation
sns.heatmap(numeric_df.corr(), annot=True)
```

**Pivot Table and Heat Map**

```python
flight_df = sns.load_dataset("flights")
pivot_table = pd.pivot_table(data=flight_df, index="month", columns="year", values="passengers")
fig = plt.figure(figsize=(15,15))
sns.heatmap(pivot_table, annot=True)
fig.savefig("heatmap.png")
```
