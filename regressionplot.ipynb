# Regression Plot
**Very Important as it is used in Machine Learning**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.lmplot(data=df, x="total_bill", y="tip")
```

```python
sns.lmplot(data=df, x="total_bill", y="tip", hue="smoker", palette="rainbow", markers=["x", "v"])
```
