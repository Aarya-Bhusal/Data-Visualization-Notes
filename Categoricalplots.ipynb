# Categorical Plots

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("taxis")
df.head(2)
```

```python
def tipped(amt):
    if amt>0:
        return "yes"
    else:
        return "no"

df["tipped"] = df["tip"].apply(tipped)
```

**Count Plot**

```python
sns.countplot(df["payment"])
```

```python
sns.countplot(x =df["payment"], hue=df["tipped"])
```

**Bar Plot**

```python
sns.barplot(df, x="pickup_borough", y="tip") # x is always qualitative and y is always quantitative. 
# By default mean of tip is shown for every individual x(qualitative data)
```

```python
# We can also change the default from mean to other functions in the y-axis
sns.barplot(data=df, x="dropoff_borough", y="tip", estimator=np.sum)
```
